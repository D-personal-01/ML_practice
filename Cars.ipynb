```python
def own(x):
    if x=="First Owner": return 1
    if x=="Second Owner": return 2
    if x=="Third Owner": return 3
    if  x=="Fourth & Above Owner": return 4
```

```python
cars["owner"] = cars["owner"].apply(own)
```

```python
import datetime

current_year = datetime.datetime.now().year
cars["age"] = current_year - cars["year"]
```

```python
cars["brand"] = cars["name"].str.split().str[0]
```

```python
brand_reliability = cars.groupby("brand")["age"].mean().sort_values(ascending=False)
```

```python
top3 = brand_reliability.head(3)

print("Top 3 Most Reliable Brands:")
print(top3)
```

```python
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

le = LabelEncoder()
cars["brand_encoded"] = le.fit_transform(cars["brand"])

plt.figure()
plt.scatter(cars["brand_encoded"], cars["age"])
plt.xlabel("Brand (Encoded)")
plt.ylabel("Age")
plt.title("Brand Reliability Distribution")
plt.show()
```
